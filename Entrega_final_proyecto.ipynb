2. Datos Demográficos:

Las variables demográficas ofrecen un contexto valioso sobre el entorno del estudiante y sus características personales. Las variables académicas son fundamentales para comprender las trayectorias educativas y el contexto académico del estudiante. # Aplicación PCA

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA


#df_final = df_final.dropna()

# Instanciar PCA
n_components = 20  # Cambia esto al número de componentes que desees
pca = PCA(n_components=n_components)

# Ajustar PCA a los datos
pca.fit(df_final)

# Transformar los datos
pca_result = pca.transform(df_final)

# Convertir a DataFrame para mejor visualización
pca_df = pd.DataFrame(data=pca_result, columns=[f'PC{i+1}' for i in range(n_components)])

# Visualizar los resultados (solo los primeros dos componentes)
plt.figure(figsize=(10, 6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], alpha=0.5)
plt.title('PCA: Proyección de Componentes Principales')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.grid()
plt.show()

# Explicación de la varianza
explained_variance = pca.explained_variance_ratio_
print("Varianza explicada por cada componente:", explained_variance)

# Opcional: Gráfico de varianza explicada acumulada
plt.figure(figsize=(10, 6))
plt.plot(range(1, n_components + 1), explained_variance.cumsum(), marker='o')
plt.title('Varianza Explicada Acumulada')
plt.xlabel('Número de Componentes')
plt.ylabel('Varianza Explicada Acumulada')
plt.grid()
plt.xticks(range(1, n_components + 1))
plt.show()

```

Primera Componente Principal: Explica aproximadamente el 11.77% de la varianza, lo que sugiere que representa una dirección importante en el espacio de los datos.

Segunda Componente Principal: Con un 9.28% de la varianza explicada, también es relevante, aunque menos que la primera.

Componentes Subsiguientes: Los siguientes componentes (hasta el octavo) explican entre 3.22% y 5.31% de la varianza, lo que muestra que aunque aportan información, su contribución es decreciente.

## K-Nearest Neighbors (KNN)

## Support Vector Machine (SVM)


```python
# Modelo SVM
svm_model = SVR(kernel='linear')  # Puedes ajustar el kernel
svm_model.fit(X_train_scaled, y_train)

# Predicciones
y_pred_svm = svm_model.predict(X_test_scaled)

# Evaluación
mae_svm = mean_absolute_error(y_test, y_pred_svm)
mse_svm = mean_squared_error(y_test, y_pred_svm)
rmse_svm = mean_squared_error(y_test, y_pred_svm, squared=False)

print("SVM:")
print(f'MAE: {mae_svm}, MSE: {mse_svm}, RMSE: {rmse_svm}')
```
